# 00 — Project Definition

## Business Problem

A retailer needs to estimate future product demand so that inventory can be
planned more effectively. Under-forecasting leads to stockouts and lost
sales; over-forecasting leads to excess inventory, storage cost, and markdowns.

**Objective:** forecast future sales using historical sales patterns and
relevant explanatory variables, then use those forecasts to drive a simple,
defensible inventory policy.

## Forecasting Unit

We forecast **daily demand for each product** (not "sales" in the abstract).
This is a panel of time series: 40 products x ~2.5 years of daily history.

```
date        product_id   demand
2023-01-01  P001         42
2023-01-02  P001         37
2023-01-03  P001         51
```

`X` = lagged demand, rolling statistics, calendar features, price/promotion
features (all known *at or before* the prediction date).
`y` = quantity sold on the target date.

## Why this is a time-series problem, not a plain regression problem

Demand on a given day is correlated with recent demand (autocorrelation),
has weekly/yearly seasonality, and the *order* of observations carries
information a plain i.i.d. regression setup ignores. That has two
consequences we treat carefully later in the project:

1. **Leakage:** features must only use information available at prediction
   time (no using tomorrow's demand to predict today's).
2. **Validation:** splits must respect time order (train on the past,
   validate on the future) rather than a random `train_test_split`.

## Roadmap

`data quality -> SQL analysis -> time-series EDA -> feature engineering ->
baseline -> ML models -> time-aware validation -> error analysis ->
inventory simulation -> business recommendation`
